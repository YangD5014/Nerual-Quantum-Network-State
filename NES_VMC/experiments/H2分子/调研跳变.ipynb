```python
import pickle
import matplotlib.pyplot as plt
import numpy as np
from H2_631G import E_fcis,K
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
```

```python
with open('./data/26-09-17-18-42_history_natural_gradient_H2_molecule_K4.pkl', 'rb') as f:
    history = pickle.load(f)
history.keys()
```

```python
fig,ax = plt.subplots(figsize=(10,8))
for i in range(4):
    ax.plot(history['steps'],np.array(history['Energy_levels'])[:,i])
    ax.hlines(E_fcis[i],0,history['steps'][-1],linestyle='--',color='k')

ax.set_ylim(-1.2,-0.2)

ax_inset = inset_axes(
    ax,
    width="40%",
    height="25%",
    loc="lower left",
    bbox_to_anchor=(0.54, 0.7, 1, 1),   # 4‑tuple！
    bbox_transform=ax.transAxes,
    borderpad=0
)

ax_inset.set_xlabel("Step", fontsize=8)
ax_inset.set_ylabel(r"$\Delta E$ (Ha)", fontsize=8)
ax_inset.set_title(f'NES-VMC (K={K}) $H_2$ Energy Levels Difference', fontsize=8)
ax_inset.tick_params(axis='both', labelsize=7)
for i in range(K):
    ax_inset.scatter(history['steps'],np.array(history['Energy_levels'])[:,i]-E_fcis[i],s=6, alpha=0.7)

# ---- 新增：0.0016以下浅蓝色填充 + 阈值虚线 ----
ax_inset.set_ylim(0.0, 0.002)
ax_inset.axhspan(ymin=0.0, ymax=0.0016, color="#b8e1ff", alpha=0.4, zorder=0)
ax_inset.axhline(y=0.0016, color="steelblue", linestyle="--", lw=0.7, zorder=1)

ax_inset.set_xlabel("Step", fontsize=8)
ax_inset.set_ylabel(r"$\Delta E$ (Ha)", fontsize=8)
ax_inset.set_title(f'NES-VMC (K={K}) $H_2$ Energy Levels Difference', fontsize=8)
ax_inset.tick_params(axis='both', labelsize=7)




ax_inset.set_xlabel("Step", fontsize=8)
ax_inset.set_ylabel(r"$\Delta E$ (Ha)", fontsize=8)
ax_inset.set_title(f'NES-VMC (K={K}) $H_2$ Energy Levels Difference', fontsize=8)
ax_inset.tick_params(axis='both', labelsize=7)


```
